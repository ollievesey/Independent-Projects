```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

class P: 
    
    RAN = [1, 0, 1, 0, 4, 3, 4, 3, 4] 
    
    def __init__(self, coeffs):                        #This creates our Polynomial class and stores the coeffecients (coeffs) in a NumPy array.
        self.coeffs = np.array(coeffs, dtype = float)
        self.trim()
        
#Method 1.1

    def evaluate(self, x):                   #Now were evaluating the polynomial for a given x.
        result = 0.                                       
        for i, a in enumerate(self.coeffs):  #Enumerate picks 'i' to act as the powers, and 'a' as the value of the coefficient stored in the array.
            result += float(a * (x ** i))
        return(float(result))

    
#Method 1.2

    def trim(self):     #Here we remove irrelevant zeros from the array of coefficients.                                            
        while len(self.coeffs) > 1 and self.coeffs[-1] == 0:    
            self.coeffs = self.coeffs[:-1]   
        return self.coeffs

    
#Method 1.3

    def plot(self, a, b):                        
        x = np.linspace(a, b, 200)          #Creates 200 points (x-values) between a and b, that are to be evaluated
        y = [self.evaluate(i) for i in x]       #Calculates f(x) for each point using the evaluate function.
        
        plt.plot(x, y)
        plt.title("Polynomial Function")
        plt.xlabel("x")
        plt.ylabel("f(x)")
        plt.grid(True)           #Adds grid to the graph.
        plt.show()           #Shows the plot (although not strictly neccesary when using Jupyter).

#Method 1.4

    def integrate(self, a, b):
        total = 0.0
        for p, q in enumerate(self.coeffs):
            total += (q*(b**(p+1) - a**(p+1)))/(p+1)  #Calculating using integration formula.
        return float(total)

    
#Method 1.5

    def deriv(self):
        self.trim()
        if len(self.coeffs) == 1: #Case for the polnomial being degree 0.
            return P([0.0])
        else:
            deriv_coeffs = [(i + 1)*float(self.coeffs[i + 1]) for i in range(len(self.coeffs) - 1)]
            return P(deriv_coeffs)

        
#Method 1.6
            

    def connect(self, coeffs2):
        C = np.array(coeffs2, dtype=float)
        new_coeffs = np.concatenate([np.array(self.RAN, dtype=float), C]) #Connects coeffs2 to RAN.
        return P(new_coeffs) #Returns the new array as a new instance new_coeffs. 
    
#Method 3.1

  
                    
    def root_poly(self, roots):
        R = np.array(roots, dtype=float)
        if len(R) > 3:
            raise ValueError("root_poly accepts at most 3 roots [a, b, c].")
        elif len(R) == 3:
            a = roots[0]
            b = roots[1]
            c = roots[2]
            a0 = -a*b*c
            a1 = a*b + b*c + c*a
            a2 = - (a + b + c)
            a3 = 1
            return P([a0, a1, a2, a3])
        elif len(R) == 2:
            a = roots[0]
            b = roots[1]
            a0 = a*b
            a1 = -(a+b)
            a2 = 1
            return P([a0, a1, a2])
        elif len(R) == 1:
            a0 = -roots[0]
            a1 = 1
            return P([a0, a1])
        

class Q(P):
      
    
    def __init__(self, coeffs):    #Quadratic subclass re-uses Polynomial methods; requires there to be <= 3 coefficients.
        super().__init__(coeffs)
        if len(coeffs) >= 4:
            print("A quadratic requires at most 3 coefficients: [a0, a1, a2].")
        
        
        
        
# Method 2.1
        
    def roots(self):
            
        if len(self.coeffs) == 3:     #The 1st case for 3 coefficients (full quadratic).
           
            a = float(self.coeffs[2])    #Saving the coefficients as there common name for implementation of the quaratic formula.
            b = float(self.coeffs[1])
            c = float(self.coeffs[0])
            
            D = b**2 - 4*a*c
            root_D = np.sqrt(D + 0j)  #This part calculates the discriminant in NumPy so it can save the result as a complex number (if neccesary).
          
            
            root_1 = (-b + root_D)/(2*a)  #Calculates and associates the (possible) two roots to variables.
            root_2 = (-b - root_D)/(2*a)
                      
            if root_1 == root_2:  #Here is the case of a repeated root.
                print("There is a repeated solution to f(x) = 0 at x = ",root_1.real) 
                
            elif D < 0:   #Here is the case that we have complex roots.
                print("The quadratic has two complex solutions to f(x) = 0; at x = ",root_1.real,"+",root_1.imag,"i", ", and x = ",root_2.real,"+",root_2.imag,"i")   #Here using .real and .imag so it prints the results more cleanly.
                
            else:   #Here is the case for two distinct real roots.
                print("The quadratic has two solutions to f(x) = 0; at x = ",root_1.real, ", and x = ",root_2.real)
                
        elif len(self.coeffs) == 2:          #The 2nd case is for solving a linear equation (2 coefficients). 
            root = -(self.coeffs[0])/self.coeffs[1]      #Calculating the root: If ax + b = 0, x = -b/a
            print("This linear equation has a solution to f(x) = 0 at x = ",root)
            
        else:
            if self.coeffs[0] == 0:    #Constant equation case; roots only exist if the constant coefficient is 0, as 0 = 0 for all x values
                print("This is a constant equation that can be represented as y = 0, thus it has infinite solutions at all x values.")
                
            else:
                print("This is a constant equation that doesn't intersect the x-axis, thus it has no solutions to f(x) = 0.")
            

#Method 2.2


    def extrema(self, c, d):
            
        
            c = float(c)  #Saving as float for better calculations.
            d = float(d) 
        
            if d <= c:
                print("It is required that c < d for the closed interval [c, d].")
            
            start = self.evaluate(c)  #Findnig the value of these end points. 
            end = self.evaluate(d)
                   
        
            if len(self.coeffs) == 3:    #Case 1 again for three coefficients (quadratic with non-zero x-coefficient).
            
                dydx_m = (self.coeffs[2])*2    #Differentiating our quadratic and saving coefficients in the form mx + c, with m non-zero). 
                dydx_c = self.coeffs[1]
           
            
                turn_point = float(-(dydx_c)/dydx_m)   #Finding the turning is solving mx + c = 0, giving x = -c/m.
            
                if c <= turn_point <= d: #Subcase for when the turning point is inside the closed interval (including end points).
                
                    extreme = self.evaluate(turn_point)  #Finding the value at the turning point.
                
                    if self.coeffs[2] > 0: #Different reports based on parabola:
                        print("This quadratic has a 'postitive' parabola, and the tunrning point is located within the closed interval.") 
                        print("Thus, the minimum point is the turning point (",turn_point,",",extreme,").")
                        
                        if start == end:
                            print("There are two maximum points at (",c,",",start,"), and (",d,",",end,").") #Multiple maximum points case.
                        elif max(start, end) == start:
                            print("The maximum point is then (",c,",",start,").")
                        else:
                            print("The maximum point is then (",d,",",end,").")
                        
                    elif self.coeffs[2] < 0:
                        print("This quadratic has a 'negative' parabola, and the tunrning point is located within the closed interval.")
                        print("Thus, the maximum point is the turning point (",turn_point,",",extreme,").")
                        
                        if start == end:
                            print("There are two minimum points at (",c,",",start,"), and (",d,",",end,").") #Multiple minmum points in case.
                        if min(start, end) == start:
                            print("The minimum point is then (",c,",",start,").")
                        else:
                            print("The minimum point is then (",d,",",end,").")
             
                elif turn_point < c:   #Different reports based on both parabola and location of turning point.
                    
                    if self.coeffs[2] > 0:
                        
                        print("This quadratic has a 'postitive' parabola, and the tunrning point is located before the closed interval.")
                        print("Thus, the minimum point is the start of the closed interval (",c,",",start,").")
                        print("The maximum point is then the end of the closed interval (",d,",",end,").")
                
                    elif self.coeffs[2] < 0:
                        print("This quadratic has a 'negative' parabola, and the tunrning point is located before the start of the closed interval.")
                        print("Thus, the maximum point is the start of the closed interval (",c,",",start,").")
                        print("The minimum point is then the end of the closed interval (",d,",",end,").")
                    
                elif turn_point > d: 
                    
                    if self.coeffs[2] > 0:
                        print("This quadratic has a 'postitive' parabola, and the tunrning point is located after the end of the closed interval.")
                        print("Thus, the maximum point is the start of the closed interval (",c,",",start,").")
                        print("The minimum point is then the end of the closed interval (",d,",",end,").")
                    
                    elif self.coeffs[2] < 0:
                        print("This quadratic has a 'negative' parabola, and the tunrning point is located after the end of the closed interval.")
                        print("Thus, the minimum point is the start of the closed interval (",c,",",start,").")
                        print("The maximum point is then the end of the closed interval (",d,",",end,").")
                    
                    
            elif len(self.coeffs) == 2: #Case 2 for two coefficients (linear with non-zero x-coefficient).
            
                if self.coeffs[1] > 0:  #Different reports based on the coefficient of the x-term being positive or negative.
                    print("This is a linear equation with the x-coefficent greater than 0.")
                    print("Thus, the minimum point is the start of the closed interval (",c,",",start,").")
                    print("The maximum point is then the end of the closed interval (",d,",",end,").")
                
                elif self.coeffs[1] < 0:
                    print("This is a linear equation with the x-coefficent less than 0.")
                    print("Thus, the maximum point is the start of the closed interval (",c,",",start,").")
                    print("The minimum point is then the end of the closed interval (",d,",",end,").")
                
            elif len(self.coeffs) == 1: #Same report for extreme values of a constant function each time.
                
                print("This is a constant function, so every point in the closed interval is a maximum and a minimum point, with value",self.coeffs[0],".")  
                

#Method 3.2

    def turning_point(self):       
               
        if len(self.coeffs) == 3:            
            dydx_m = (self.coeffs[2])*2    #Differentiating our quadratic and saving coefficients in the form mx + c, with m non-zero). 
            dydx_c = self.coeffs[1]
            
            turn_pointx = float(-(dydx_c)/dydx_m)  #Finding the turning is solving mx + c = 0, giving x = -c/m.
            turn_pointy = self.evaluate(turn_pointx)
            
            print("The turning point of the quadratic function represented by the instnace is (",turn_pointx,",",turn_pointy,").")
        else:
            raise ValueError("Instance must be a polynomial representing a proper quadratic function with x^2 coefficient non-zero.")
        

       
```

## Quadratic Subclass (Q)

The Quadratic class is a child class that inherits all methods from the parent Polynomial class, such as integrate(),  plot(), etc. but is designed specifically for quadratic functions. It can then use new added analytical methods that only apply to quadratics, demonstrating the advantage of inheritance.

A quadratic function can be written as:

$f(x) = a_0 + a_1x + a_2x^2$

This class automatically checks that no more than three coefficients are entered and then adds two special methods for solving and analysing quadratic functions, roots() and extrema(c, d). There is no need for a function to evaluate $f(x)$ at a given point in the subclass, as you can simply call evaluate(x) from the parent class P.

## Method 1 — Finding the Roots

The roots (or solutions) of a quadratic function are the values of $x$, for which $f(x) = 0$.
The roots() method applies the quadratic formula:

$$ x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a} $$

where $a = a_2, b = a_1$ and $c = a_0$. Depending on the discriminant, $D = b^2 -4ac$ , the program the prints the root(s), given one of three possible cases:

- Two distinct real roots ($D > 0$)

- One repeated real root ($D = 0$)

- Two complex roots ($D < 0$)

## Method 2 — Finding Extrema on a Closed Interval

Quadratic functions form parabolas, which either open upwards or downwards depending on the sign of $a_2$:

- If $a_2 > 0$, the parabola opens upwards and has a minimum point.

- If $a_2 < 0$, the parabola opens downwards and has a maximum point.

The turning point occurs where the derivative $f'(x)$ is zero. Differentiating gives: $$f'(x) = 2a_2x + a_1$$

and setting this equal to zero gives the x-coordinate of the turning point, $$x_t = -\frac{a_1}{2a_2}$$

The program then evaluates $f(x)$ at both end points of the interval $[c, d]$, finds the turning point, checks if it lies inside the interval, and prints a detailed report identifying:

 - Whether the parabola is positive or negative,

 - Whether the turning point is inside the closed interval or not,

 - Which points represent the maximum and minimum values (including the case for multiple).
 
 

 I chose to 'print' these results rather than 'return' them as at the time, there is no need to store their results, and (to me) the problem indicates towards printing a report rather than returning a value.



## Method 3.1 - root_poly

This method accepts at most 3 roots in an NumPy array (for easy calculation) and then has three cases, relating to the following formula:

For roots $a_1, a_2, ..., a_n$ of the polynomial $f(x)$, $f(x)$ can be written as: $$ f(x) = (x-a_1)(x-a_2)...(x-a_n)$$   

By expanding these brackets, we can find the coefficients of a polynomial function with these (at most 3) roots, a, b and c:
- Three roots gives $f(x) = x^3 − (a + b + c)x^2 + (ab + bc + ca)x − abc$,
- Two roots gives $f(x) = x^2 - (a+b)x + ab$,
- One root gives $f(x) = x - a$

To run the root_poly function, you must first create an arbitrary instance of the class. Then you can call the root_poly function (inputing at most 3 roots) on this instance which will then creates a new instance. This new instance will have the coefficients of a polynomial that has roots to the ones entered in root_poly. If more than 3 roots are entered a ValueError is displayed.

## Method 3.2 - turning_point

This function checks to see wether the quadratic is 'proper' - ensuring the coefficient of $x^2$ is non-zero. It does this by checking the length of 'self.coeffs' as this is list automatically trimmed of irrelevant zero-coefficients when the values are first saved in the class constructor. 

So our original quadratic can be written of the form, $$f(x) = a_0 + a_1x + a_2x^2, a_2 \neq 0$$

The turning point occurs where the derivative $f'(x)$ is zero. Differentiating gives: $$f'(x) = 2a_2x + a_1$$

As before, setting this equal to zero gives the x-coordinate of the turning point, $$x_t = -\frac{a_1}{2a_2}$$

We then conclude the method by calculating the y-value of the turning point by calling evaluate on this x-value, and printing the final answer in a print statement.
